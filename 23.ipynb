# 23. Diverse instruments

**Question from the meeting:** does the model generalise better if it is trained on a wider and more varied set of stocks, not only large NASDAQ names?

**Design**
- Holdouts stay the same as before: XOM and JPM are never used for training.
- Training pool 0 (baseline): the 8 original tickers (SPY, QQQ, NVDA, GOOGL, AAPL, MSFT, AMZN, META, AAPL included).
- Stage A: pool 0 plus 6 more large NASDAQ stocks (AVGO, TSLA, COST, ADBE, AMD, NFLX).
- Stage B: stage A plus Intel and 7 value-style / non-NASDAQ stocks (INTC, CSCO, JNJ, KO, WMT, PG, CAT, F).
- Same model for every pool (Random Forest, 300 trees, depth 8, balanced classes), 3 random seeds each, so we can see how much of any difference is just noise.

**What would count as an improvement:** a gain larger than the seed-to-seed spread, on both holdouts.
